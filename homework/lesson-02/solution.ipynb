## 7. Самостоятельная практика «Итог заказа»

**Исходные правила:** скидка 10% при сумме от 7000; скидка 5% при сумме
от 5000 и наличии клубной карты; иначе скидка 0%. Доставка бесплатна, если
сумма после скидки не меньше 5000, иначе доставка стоит 300 рублей.

**Что сделать:**

1. Создайте `product_name`, `price`, `quantity`, `is_member`.
2. Рассчитайте `order_total`.
3. Через `if` / `elif` / `else` присвойте `discount_rate`.
4. Рассчитайте `total_after_discount`.
5. Через второе условие присвойте `actual_delivery_cost`.
6. Рассчитайте `final_total` и выведите все результаты одной f-строкой.
7. Проверьте три сценария: без скидки; со скидкой 5%; ровно 7000 до скидки.
8. Для каждого сценария заполните подготовленное поле ниже.

**Готово, если:** код выполняется без ошибок, значения меняются внутри
ветвей, проверены три сценария и вывод соответствует результату.

````python
#Сценарий 1: без скидки

# 1. Исходные данные

product_name = "Попрыгунчик"
price = 1000
quantity = 2
is_member = False

# 2. Исходная сумма

order_total = price * quantity

# 3. Выбор discount_rate

if (order_total >= 7000):
  discount_rate = 0.1
elif((order_total >= 5000) and (is_member == True)):
  discount_rate = 0.05
else:
  discount_rate = 0

# 4. Сумма после скидки

total_after_discount = order_total * (1 - discount_rate)

# 5. Выбор actual_delivery_cost

if(total_after_discount >= 5000):
  actual_delivery_cost = 0
else:
  actual_delivery_cost = 300

# 6. Итоговая сумма и f-строка

final_total = total_after_discount + actual_delivery_cost
print(f"Товар: {product_name};\n"
      f"Количество: {quantity};\n"
      f"Цена за шт.: {price};\n"
      f"Наличие клубной карты: {is_member};\n"
      f"Сумма покупки без учета скидки: {order_total};\n"
      f"Скидка: {discount_rate};\n"
      f"Сумма покупки с учетом скидки: {total_after_discount};\n"
      f"Стоимость доставки: {actual_delivery_cost};\n"
      f"Сумма покупки с учетом скидки и стоимости доставки: {final_total};")
````

## Вопрос и гипотеза

**Мой вопрос:** По итогам напсианяи и выполненяи программы будет ли применена скидка и какова будет стоимость доставки?
**Моя гипотеза:**  Я предполагаю, что при цене товара 1400 руб. и количестве 3 шт. начальная сумма заказа составит 4200 руб. Так как эта сумма меньше порога для скидки (5000 руб.), даже несмотря на наличие клубной карты и меньше порога для бесплатной доставки (6000 руб.), скидка не будет применена (0%), а к сумме добавится стоимость доставки (400 руб.). Итоговая сумма (final_total) должна составить 4600 рублей.
